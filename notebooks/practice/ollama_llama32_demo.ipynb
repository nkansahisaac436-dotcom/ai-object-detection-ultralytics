# 🦙 Running Llama 3.2 (1B) Locally with Ollama and Python

Welcome! In this practice notebook, we will learn how to interact with an AI model running directly on your own computer.

### 💡 What is Ollama?
**Ollama** is a free, open-source tool that allows you to download and run Large Language Models (LLMs) locally on your PC. It does not send your private data to any remote cloud server.

### 💡 What is Llama 3.2 (1B)?
**Llama 3.2 (1B)** is a lightweight, highly capable language model created by Meta. With approximately 1 billion parameters, it is designed to run smoothly and quickly on everyday computer processors (**CPU**) without needing an expensive dedicated GPU.

## Step 1: Install and Import the Ollama Python Library

To connect to Ollama from Python, we use the official `ollama` package:
```bash
pip install ollama
```

````python
import ollama

print("Ollama Python library imported successfully!")
````

## Step 2: List Local Models

Let's check which models are currently downloaded and ready to use on our local machine.

````python
response = ollama.list()
print("Installed local models:")
for m in response.get('models', []):
    name = getattr(m, 'model', None) or (m.get('name') if isinstance(m, dict) else str(m))
    print(f" - {name}")
````

## Step 3: Ask a Simple Question (Single Prompt)

We use `ollama.chat()` to send a question to the model and receive a complete answer.

````python
prompt = "What is artificial intelligence in one simple sentence?"

response = ollama.chat(
    model='llama3.2:1b',
    messages=[
        {'role': 'user', 'content': prompt}
    ]
)

print(f"Question: {prompt}\n")
print("Answer:")
print(response['message']['content'])
````

## Step 4: Stream the Response in Real Time

Streaming prints words as they are being generated, giving a fast interactive experience just like ChatGPT.

````python
prompt = "Give 3 beginner tips for learning Python programming."
print(f"Question: {prompt}\n")
print("Answer (streaming):\n")

stream = ollama.chat(
    model='llama3.2:1b',
    messages=[{'role': 'user', 'content': prompt}],
    stream=True
)

for chunk in stream:
    print(chunk['message']['content'], end='', flush=True)
print()
````

## Step 5: Multi-Turn Chat (Memory & Conversation History)

Language models do not automatically remember past messages. To have a conversation, we keep a list of messages and append each new reply.

````python
# 1. Initialize conversation with user message
history = [
    {'role': 'user', 'content': 'Hello! My name is Isaac and I am studying AI at Pentecost University.'}
]

# 2. Get first response
reply_1 = ollama.chat(model='llama3.2:1b', messages=history)
print("User: Hello! My name is Isaac and I am studying AI at Pentecost University.")
print(f"AI:   {reply_1['message']['content']}\n")

# 3. Append assistant response and ask a follow-up
history.append(reply_1['message'])
history.append({'role': 'user', 'content': 'What is my name and what university do I attend?'})

# 4. Get second response using full history
reply_2 = ollama.chat(model='llama3.2:1b', messages=history)
print("User: What is my name and what university do I attend?")
print(f"AI:   {reply_2['message']['content']}")
````

## 🎉 Key Takeaways
- **Privacy & Speed:** Ollama runs 100% locally on your PC CPU without requiring an internet connection.
- **Simple Python Interface:** `ollama.chat(model=..., messages=...)` is easy to integrate into any AI project.
- **Streaming & Context:** Supports streaming tokens and maintaining conversation histories.
