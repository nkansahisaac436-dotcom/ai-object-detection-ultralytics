## Python Basics: Variables and Operations

### Variables
In Python, a variable is a named storage location for data. You can think of it as a container that holds a value. Variables are created when you assign a value to them using the `=` operator. Python is dynamically typed, meaning you don't need to declare the variable's type explicitly; the interpreter infers it from the assigned value.

```python
# Assigning different types of values to variables

# Integer variable
my_integer = 10
print(f"my_integer: {my_integer}, Type: {type(my_integer)}")

# Float variable
my_float = 20.5
print(f"my_float: {my_float}, Type: {type(my_float)}")

# String variable
my_string = "Hello, Python!"
print(f"my_string: {my_string}, Type: {type(my_string)}")

# Boolean variable
is_true = True
print(f"is_true: {is_true}, Type: {type(is_true)}")

# Reassigning a variable
my_integer = 15
print(f"my_integer after reassignment: {my_integer}")
```

### Basic Operations

Once you have variables, you can perform various operations on them. These include arithmetic operations, comparison operations, and logical operations.

```python
# Arithmetic Operations
x = 10
y = 3

print(f"Addition (x + y): {x + y}")
print(f"Subtraction (x - y): {x - y}")
print(f"Multiplication (x * y): {x * y}")
print(f"Division (x / y): {x / y}") # Float division
print(f"Floor Division (x // y): {x // y}") # Integer division
print(f"Modulo (x % y): {x % y}") # Remainder
print(f"Exponentiation (x ** y): {x ** y}")

# String Concatenation
str1 = "Hello"
str2 = "World"
print(f"String Concatenation: {str1 + ' ' + str2}")

# Comparison Operations
a = 5
b = 10

print(f"Equal to (a == b): {a == b}")
print(f"Not equal to (a != b): {a != b}")
print(f"Greater than (a > b): {a > b}")
print(f"Less than (a < b): {a < b}")
print(f"Greater than or equal to (a >= b): {a >= b}")
print(f"Less than or equal to (a <= b): {a <= b}")

# Logical Operations
p = True
q = False

print(f"AND (p and q): {p and q}")
print(f"OR (p or q): {p or q}")
print(f"NOT (not p): {not p}")
```

## Loops

Loops are fundamental control structures in programming that allow you to execute a block of code multiple times. Python provides two main types of loops: `for` loops and `while` loops.

### The `for` Loop

The `for` loop is used for iterating over a sequence (like a list, tuple, string, or range) or other iterable objects. It executes a block of code once for each item in the sequence.

```python
# Example 1: Iterating over a list
fruits = ["apple", "banana", "cherry"]
print("Iterating over a list:")
for fruit in fruits:
    print(fruit)

# Example 2: Iterating over a string
my_word = "Python"
print(
"\nIterating over a string:")
for char in my_word:
    print(char)

# Example 3: Using range() to loop a specific number of times
print("\nLooping a specific number of times (0 to 4):")
for i in range(5):
    print(i)

# Example 4: Using range() with start, stop, and step
print("\nLooping with start, stop, and step (2 to 10, step 2):")
for i in range(2, 11, 2):
    print(i)
```

### The `while` Loop

The `while` loop repeatedly executes a block of code as long as a given condition is true. It's suitable when you don't know the exact number of iterations beforehand.

```python
# Example 1: Basic while loop
count = 0
print("Basic while loop:")
while count < 5:
    print(f"Count is: {count}")
    count += 1 # Increment count to eventually make the condition false

# Example 2: while loop with a break statement
num = 0
print("\nWhile loop with break:")
while True:
    print(f"Current number: {num}")
    if num >= 3:
        break # Exit the loop when num is 3 or more
    num += 1

# Example 3: while loop with a continue statement
# Prints only odd numbers less than 10
print("\nWhile loop with continue (printing odd numbers):")
i = 0
while i < 10:
    i += 1
    if i % 2 == 0: # If i is even, skip the rest of the loop body and go to the next iteration
        continue
    print(f"Odd number: {i}")
```

## Classes

In Python, a class is a blueprint for creating objects (instances). It's a way to bundle data (attributes) and functionality (methods) together. Classes allow you to implement Object-Oriented Programming (OOP) concepts, promoting code reusability and organization.

### Defining a Class

A class is defined using the `class` keyword. Inside the class, you typically define:
- **Attributes**: Variables that store data associated with an object. They can be class-level (shared by all instances) or instance-level (unique to each instance).
- **Methods**: Functions defined inside a class that perform actions on the object's data. The first parameter of any instance method is conventionally named `self`, which refers to the instance of the class itself.
- **`__init__` method**: A special method called the constructor. It's automatically called when a new object of the class is created. It's used to initialize the object's attributes.

```python
# Define a simple class named 'Dog'
class Dog:
    # Class attribute (shared by all Dog instances)
    species = "Canis familiaris"

    # The constructor method: initializes attributes when a new Dog object is created
    def __init__(self, name, age):
        self.name = name  # Instance attribute
        self.age = age    # Instance attribute

    # Instance method: an action a Dog object can perform
    def bark(self):
        return f"{self.name} says Woof!"

    # Another instance method
    def description(self):
        return f"{self.name} is {self.age} years old."

# Creating objects (instances) of the Dog class
my_dog = Dog("Buddy", 3)
your_dog = Dog("Lucy", 5)

# Accessing attributes
print(f"My dog's name is {my_dog.name}.")
print(f"Your dog's age is {your_dog.age}.")
print(f"All dogs are of species: {Dog.species}") # Accessing class attribute

# Calling methods
print(my_dog.bark())
print(your_dog.description())

# You can also change attributes after creation
my_dog.age = 4
print(f"My dog's new age is {my_dog.age}.")
```

## Functions

A function is a block of organized, reusable code that performs a single, related action. Functions provide better modularity for your application and a high degree of code reusing. Python has many built-in functions, and you can also define your own functions.

### Defining and Calling Functions

- **Defining a function**: You define functions using the `def` keyword, followed by the function name, parentheses `()`, and a colon `:`. The code block for the function is indented.
- **Calling a function**: You execute a function by using its name followed by parentheses `()`.
- **Parameters and Arguments**: You can pass information into functions as parameters. Parameters are specified inside the parentheses in the function definition. When you call the function, the values passed are called arguments.
- **Return Values**: A function can return data as a result. The `return` statement is used for this purpose.

```python
# Example 1: A simple function without parameters or return value
def greet():
    print("Hello there!")

# Call the function
greet()

# Example 2: Function with parameters
def greet_person(name):
    print(f"Hello, {name}!")

greet_person("Alice")
greet_person("Bob")

# Example 3: Function with a return value
def add(a, b):
    result = a + b
    return result

sum_result = add(5, 3)
print(f"The sum of 5 and 3 is: {sum_result}")

# Example 4: Function with default parameter values
def say_hello(name="Guest", greeting="Hello"):
    print(f"{greeting}, {name}!")

say_hello() # Uses default values
say_hello("Charlie") # Overrides 'name'
say_hello("David", "Hi") # Overrides both

# Example 5: Function with multiple return values (as a tuple)
def get_name_parts(full_name):
    parts = full_name.split()
    first_name = parts[0]
    last_name = parts[-1]
    return first_name, last_name

first, last = get_name_parts("John Doe")
print(f"First name: {first}, Last name: {last}")
```

## Modules and Importing Modules

### What are Modules?
A **module** is simply a file containing Python code (like functions, classes, or variables). It helps you organize your code into reusable units, making large programs easier to manage and understand. Think of a module as a toolbox where each tool does a specific job.

### Importing Modules
To use code from a module, you need to `import` it into your current script. Here are the two most common ways:

1.  **`import module_name`**: Imports the whole module. You then access its contents using `module_name.item_name`.
2.  **`from module_name import item_name`**: Imports specific items (functions, classes, variables) directly. You can then use `item_name` without the module prefix.

```python
# Example 1: Importing a built-in module (math)

import math

# Use a function from the math module
print(f"The value of pi is: {math.pi}")
print(f"Square root of 25 is: {math.sqrt(25)}")

# Example 2: Importing a specific item from a module
from datetime import date

today = date.today()
print(f"Today's date is: {today}")
```
