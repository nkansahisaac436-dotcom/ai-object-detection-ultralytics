## String Formatting in `print()` Statements

Python offers several ways to format strings, allowing you to embed variables and expressions directly into a string literal. Here are the most common methods:

1.  **F-strings (Formatted string literals)**: Introduced in Python 3.6, these offer a concise and readable way to embed expressions inside string literals. You prefix the string with `f` or `F`.

2.  **`str.format()` method**: This method allows you to create a formatted string using placeholders `{}` within the string and then calling `.format()` with the values to be inserted.

3.  **Old-style `%` formatting**: Similar to C's `printf` function, this uses `%` operators to format strings. While still functional, it's generally less preferred than f-strings or `str.format()` for new code due to readability and potential errors.

```python
# 1. F-strings (Formatted string literals - Recommended for Python 3.6+)
name = "Charlie"
score = 95.789
is_passed = True

print(f"Hello, {name}! Your score is {score:.2f}.") # .2f formats to 2 decimal places
print(f"Did {name} pass? {is_passed}.")
print(f"Calculation: {2 * 3 + 1}") # Can embed expressions

print("\n---\n") # Separator

# 2. str.format() method
item = "book"
quantity = 3
unit_price = 15.99

print("You bought {} {}(s) for ${:.2f}.".format(quantity, item, quantity * unit_price))
print("Using named arguments: {item_name} costs ${price:.2f}.".format(item_name=item, price=unit_price))

print("\n---\n") # Separator

# 3. Old-style % formatting (less common in new code)
product_id = 101
product_name = "Keyboard"
product_weight = 0.75

print("Product ID: %d, Name: %s, Weight: %.2f kg" % (product_id, product_name, product_weight))
```

## Function for Formatted Mathematical Expressions

This function takes two numbers and an operator as input, performs the calculation, and then prints the entire expression along with its result using f-string formatting.

```python
def calculate_and_print(num1, num2, operator):
  """Calculates a mathematical expression and prints the formatted result."""
  result = None
  if operator == '+':
    result = num1 + num2
  elif operator == '-':
    result = num1 - num2
  elif operator == '*':
    result = num1 * num2
  elif operator == '/':
    if num2 != 0:
      result = num1 / num2
    else:
      print(f"Error: Division by zero is not allowed for {num1} {operator} {num2}")
      return
  else:
    print(f"Error: Invalid operator '{operator}'. Please use +, -, *, or /.")
    return

  print(f"The result of {num1} {operator} {num2} is: {result}")

# Example usage of the calculation function
calculate_and_print(10, 5, '+')
calculate_and_print(20, 4, '/')
calculate_and_print(7, 3, '*')
calculate_and_print(15, 8, '-')
calculate_and_print(10, 0, '/') # Example of error handling
calculate_and_print(5, 2, '%') # Example of invalid operator
```

### Comparison Operations

Comparison operators are used to compare two values. They return `True` or `False` depending on the condition.

```python
# Define some variables for comparison
a = 10
b = 20
c = 10

print(f"a = {a}, b = {b}, c = {c}")

# Equal to (==)
print(f"a == b: {a == b}") # False
print(f"a == c: {a == c}") # True

# Not equal to (!=)
print(f"a != b: {a != b}") # True
print(f"a != c: {a != c}") # False

# Greater than (>)
print(f"a > b: {a > b}") # False
print(f"b > a: {b > a}") # True

# Less than (<)
print(f"a < b: {a < b}") # True
print(f"b < a: {b < a}") # False

# Greater than or equal to (>=)
print(f"a >= b: {a >= b}") # False
print(f"a >= c: {a >= c}") # True

# Less than or equal to (<=)
print(f"a <= b: {a <= b}") # True
print(f"a <= c: {a <= c}") # True
```

### Logical Operations

Logical operators are used to combine conditional statements.

```python
# Define some boolean variables for logical operations
x = True
y = False
z = True

print(f"x = {x}, y = {y}, z = {z}")

# AND operator (and)
# Returns True if both statements are true
print(f"x and y: {x and y}") # False
print(f"x and z: {x and z}") # True

# OR operator (or)
# Returns True if at least one of the statements is true
print(f"x or y: {x or y}") # True
print(f"y or y: {y or y}") # False

# NOT operator (not)
# Reverses the result, returns False if the result is true
print(f"not x: {not x}") # False
print(f"not y: {not y}") # True

# Combining logical and comparison operations
age = 25
has_license = True

print(f"\nCan drive (age > 18 and has_license): {age > 18 and has_license}") # True

score = 85
attendance = 90

print(f"Passed course (score > 70 or attendance > 80): {score > 70 or attendance > 80}") # True
```

## Classes and Objects

In Python, **Object-Oriented Programming (OOP)** is a programming paradigm that uses "objects" – data structures consisting of data fields and methods – to design applications and computer programs. A **class** is a blueprint for creating objects, providing initial values for state (member variables or attributes) and implementations of behavior (member functions or methods).

### Defining a Simple Class

Let's define a simple class called `Dog` with some attributes and a method.

```python
class Dog:
  # The __init__ method is a special method called a constructor.
  # It's automatically called when a new object of the class is created.
  def __init__(self, name, breed, age):
    self.name = name  # Attribute: name of the dog
    self.breed = breed # Attribute: breed of the dog
    self.age = age    # Attribute: age of the dog

  # A method (function inside a class) that describes the dog's action
  def bark(self):
    return f"{self.name} says Woof! Woof!"

  def describe(self):
    return f"{self.name} is a {self.age}-year-old {self.breed}."

# Creating objects (instances) of the Dog class
my_dog = Dog("Buddy", "Golden Retriever", 3)
her_dog = Dog("Lucy", "Labrador", 5)

# Accessing attributes
print(f"My dog's name is {my_dog.name}.")
print(f"Her dog's breed is {her_dog.breed}.")

# Calling methods
print(my_dog.bark())
print(her_dog.describe())
```

### Inheritance in Python

**Inheritance** is a powerful feature in OOP that allows us to define a class that inherits all the methods and properties from another class. The class that inherits is called the **child class** or **subclass**, and the class it inherits from is called the **parent class** or **superclass**.

**Benefits of Inheritance:**
*   **Code Reusability**: You can reuse code from the parent class without rewriting it.
*   **Logical Organization**: It helps organize code into a more logical, hierarchical structure.
*   **Extensibility**: You can extend the functionality of existing classes without modifying them.

#### Example: Creating a `WorkingDog` Subclass

Let's create a new class called `WorkingDog` that inherits from our `Dog` class. The `WorkingDog` will have all the properties of a `Dog` but can also have additional attributes and methods specific to working dogs, like a `task`.

```python
class WorkingDog(Dog): # WorkingDog inherits from Dog
  def __init__(self, name, breed, age, task):
    # Call the constructor of the parent class (Dog)
    super().__init__(name, breed, age)
    self.task = task # Add a new attribute specific to WorkingDog

  # Add a new method specific to WorkingDog
  def perform_task(self):
    return f"{self.name} is performing its task: {self.task}."

  # You can also override methods from the parent class
  def bark(self):
    return f"{self.name} barks attentively while {self.task}!"

# Creating an object of the WorkingDog subclass
shepherd = WorkingDog("Max", "German Shepherd", 4, "guarding")

# Accessing inherited attributes from the Dog class
print(f"Working dog's name: {shepherd.name}")
print(f"Working dog's breed: {shepherd.breed}")
print(f"Working dog's age: {shepherd.age}")

# Accessing its own specific attribute
print(f"Working dog's task: {shepherd.task}")

# Calling inherited methods (some may be overridden)
print(shepherd.describe()) # Inherited and works as before
print(shepherd.bark()) # Overridden method

# Calling its own specific method
print(shepherd.perform_task())
```

### Overriding vs. Extending Methods

When working with inheritance, you'll often encounter situations where a subclass needs to modify or add to the behavior of a method inherited from its parent class. This leads to two main concepts: **method overriding** and **method extension**.

#### Method Overriding

**Overriding** a method means providing a completely new implementation for a method in the subclass that already exists in the superclass. When the method is called on an object of the subclass, the subclass's version is executed instead of the superclass's version.

*   **Example from `WorkingDog`**: The `bark()` method in `WorkingDog` completely replaces the `bark()` method from the `Dog` class.

```python
print("--- Method Overriding Example ---")

# Original Dog bark method
my_dog = Dog("Buddy", "Golden Retriever", 3)
print(f"Original bark: {my_dog.bark()}")

# Overridden WorkingDog bark method
shepherd = WorkingDog("Max", "German Shepherd", 4, "guarding")
print(f"Overridden bark: {shepherd.bark()}")
```

#### Method Extension

**Extending** a method means adding new functionality to an inherited method while still retaining and calling the original functionality from the superclass. This is typically done by calling the parent class's method using `super()` within the subclass's method, and then adding more code.

*   **Example**: Let's modify our `WorkingDog` to extend the `describe()` method.

```python
class EnhancedWorkingDog(Dog):
  def __init__(self, name, breed, age, task):
    super().__init__(name, breed, age)
    self.task = task

  # This method extends the 'describe' method from the Dog class
  def describe(self):
    # Call the original describe method from the parent class
    parent_description = super().describe()
    return f"{parent_description} And {self.name} is specifically trained for {self.task}."

  def bark(self):
    return f"{self.name} barks attentively while {self.task}!"

  def perform_task(self):
    return f"{self.name} is performing its task: {self.task}."


print("\n--- Method Extension Example ---")

# Creating an object of the EnhancedWorkingDog subclass
rescue_dog = EnhancedWorkingDog("Rex", "St. Bernard", 6, "rescue operations")

# Calling the extended describe method
print(rescue_dog.describe())
```

### Using `super()` to Access Parent Attributes (via `__init__`)

The most common use case for `super()` to interact with parent attributes is within the subclass's `__init__` method. When you create a subclass, its `__init__` method often needs to ensure that the parent class's `__init__` method is also called to properly initialize the attributes inherited from the parent.

**Syntax:** `super().__init__(*args, **kwargs)`

This call executes the parent class's constructor, ensuring that all attributes defined in the parent's `__init__` are set up correctly before the subclass adds its own specific attributes.

```python
class Animal:
  def __init__(self, species, sound):
    self.species = species
    self.sound = sound

  def make_sound(self):
    return f"The {self.species} says {self.sound}."

class Cat(Animal):
  def __init__(self, name, age, sound="Meow"):
    # Call the parent class's __init__ method using super()
    # This initializes 'species' and 'sound' from the Animal class
    super().__init__("Cat", sound)
    self.name = name # New attribute specific to Cat
    self.age = age   # New attribute specific to Cat

  def describe_cat(self):
    # 'species' is an attribute inherited and initialized by the parent class via super()
    return f"{self.name} is a {self.age}-year-old {self.species}."

# Create an instance of the Cat subclass
my_cat = Cat("Whiskers", 2)

# Accessing attributes initialized by the parent class
print(f"My cat's species: {my_cat.species}")
print(f"My cat's sound: {my_cat.sound}")

# Accessing attributes specific to the Cat class
print(f"My cat's name: {my_cat.name}")
print(f"My cat's age: {my_cat.age}")

# Calling inherited method
print(my_cat.make_sound())

# Calling subclass-specific method
print(my_cat.describe_cat())
```

As you can see, `super().__init__('Cat', sound)` in the `Cat` class's constructor is responsible for setting up the `species` and `sound` attributes which come from the `Animal` class. Without this call, those attributes would not be defined in a `Cat` object unless explicitly set within the `Cat`'s `__init__` method, defeating the purpose of inheritance.

### Multiple Inheritance

**Multiple Inheritance** is a feature in which a class can inherit attributes and methods from more than one parent class. This means a single child class can be a specialized version of multiple general classes, incorporating functionalities from all of them.

**Example**: Let's create a `FlyingCar` class that inherits from both a `Car` class and an `Aircraft` class.

```python
class Car:
  def __init__(self, brand, model):
    self.brand = brand
    self.model = model
    self.speed = 0

  def accelerate(self, increment):
    self.speed += increment
    return f"The {self.brand} {self.model} is now driving at {self.speed} km/h."

  def brake(self, decrement):
    self.speed = max(0, self.speed - decrement)
    return f"The {self.brand} {self.model} slowed down to {self.speed} km/h."

class Aircraft:
  def __init__(self, wingspan, max_altitude):
    self.wingspan = wingspan
    self.max_altitude = max_altitude
    self.altitude = 0

  def take_off(self):
    self.altitude = 100 # Initial altitude after takeoff
    return f"Aircraft is taking off. Current altitude: {self.altitude} meters."

  def land(self):
    self.altitude = 0
    return f"Aircraft is landing. Current altitude: {self.altitude} meters."

class FlyingCar(Car, Aircraft): # Inherits from both Car and Aircraft
  def __init__(self, brand, model, wingspan, max_altitude):
    # Call constructors of both parent classes using super()
    super().__init__(brand, model) # Initializes Car attributes (brand, model, speed)
    Aircraft.__init__(self, wingspan, max_altitude) # Initializes Aircraft attributes (wingspan, max_altitude, altitude)

  def fly(self, target_altitude):
    if target_altitude <= self.max_altitude:
      self.altitude = target_altitude
      return f"The flying car is now flying at {self.altitude} meters."
    else:
      return f"Cannot fly above max altitude of {self.max_altitude} meters."

# Create an instance of the FlyingCar
my_flying_car = FlyingCar("Terrafugia", "Transition", 8, 3000)

print(f"Flying car: {my_flying_car.brand} {my_flying_car.model}")

# Use methods from Car class
print(my_flying_car.accelerate(50))
print(my_flying_car.brake(20))

# Use methods from Aircraft class
print(my_flying_car.take_off())
print(my_flying_car.fly(1500))
print(my_flying_car.land())

# Access attributes from both parent classes
print(f"Wingspan: {my_flying_car.wingspan} meters")
print(f"Current speed: {my_flying_car.speed} km/h")
```

## Modules in Python

### Creating and Importing a Custom Module

To create your own module, you simply need to save Python code in a file with a `.py` extension. This file can contain functions, classes, variables, or any other valid Python code.

Once created, you can import this module into other Python scripts or notebooks using the `import` statement, just like you would with built-in modules.

```python
# Step 1: Create a custom module file
# We'll write some code into a file named 'my_custom_module.py'

module_content = """
def greet(name):
  return f"Hello, {name}! Welcome to your custom module."

def add_one(number):
  return number + 1

MY_FAVORITE_NUMBER = 42
"""

# Write the content to a file
with open('my_custom_module.py', 'w') as f:
    f.write(module_content)

print("Created 'my_custom_module.py' with sample functions and a variable.")
```

```python
# Step 2: Import and use the custom module

# You can import the entire module
import my_custom_module

# Access functions and variables using the module name
message = my_custom_module.greet("Learner")
print(message)

result = my_custom_module.add_one(10)
print(f"10 + 1 from module: {result}")

print(f"My favorite number from module: {my_custom_module.MY_FAVORITE_NUMBER}")

# You can also import specific items from the module
from my_custom_module import greet, MY_FAVORITE_NUMBER

print(greet("Advanced Learner"))
print(f"Direct access to MY_FAVORITE_NUMBER: {MY_FAVORITE_NUMBER}")

# If you modify the module file and want to see changes without restarting the kernel,
# you can reload it (useful during development):
import importlib
importlib.reload(my_custom_module)

print("\nCustom module demonstrated successfully!")
```

### Using `if __name__ == "__main__":` in a Module

The `if __name__ == "__main__":` block is a common idiom in Python scripts that allows you to determine whether the script is being run directly or being imported as a module into another script.

*   When a Python script is run directly (e.g., `python your_script.py`), the special variable `__name__` is set to `"__main__"`.
*   When a Python script is imported as a module into another script, the `__name__` variable is set to the module's name (e.g., `"my_custom_module"`).

This construct is useful for including code that should only execute when the module is run as the main program, such as:
*   Unit tests
*   Command-line argument parsing
*   Demonstration code or examples

Let's modify our `my_custom_module.py` to include this block and see it in action.

```python
# Step 1: Modify the custom module file to include `if __name__ == "__main__":`

module_content_with_main = r"""
def greet(name):
  return f"Hello, {name}! Welcome to your custom module."

def add_one(number):
  return number + 1

MY_FAVORITE_NUMBER = 42

if __name__ == "__main__":
  print("\n--- Running my_custom_module.py directly ---")
  print(greet("Direct Runner"))
  print(f"Adding one to 5: {add_one(5)}")
  print(f"Accessing favorite number directly: {MY_FAVORITE_NUMBER}")
  print("-----------------------------------------")
"""

# Overwrite the content of the file with the new version
with open('my_custom_module.py', 'w') as f:
    f.write(module_content_with_main)

```

## Organizing Code into Multiple Files

When your Python project grows, keeping all your code in a single file becomes unmanageable. By splitting code into multiple `.py` files (modules), you can:

*   **Improve readability and organization**: Each file can focus on a specific part of your application.
*   **Promote reusability**: Functions, classes, or variables defined in one module can be easily imported and used in others.
*   **Simplify debugging**: Isolating issues to smaller, more focused files.
*   **Facilitate collaboration**: Multiple developers can work on different files simultaneously without conflicts.

Let's create three files: `math_operations.py`, `text_utils.py`, and `main_application.py`. The `main_application.py` will import and use functions from the other two.

```python
# 1. Create 'math_operations.py' module
math_content = """
def add(a, b):
  return a + b

def subtract(a, b):
  return a - b

def multiply(a, b):
  return a * b

def divide(a, b):
  if b == 0:
    raise ValueError("Cannot divide by zero!")
  return a / b

if __name__ == '__main__':
  print("Running math_operations.py directly:")
  print(f"5 + 3 = {add(5, 3)}")
"""

with open('math_operations.py', 'w') as f:
    f.write(math_content)
print("Created 'math_operations.py'")

# 2. Create 'text_utils.py' module
text_content = """
def capitalize_string(text):
  return text.capitalize()

def reverse_string(text):
  return text[::-1]

if __name__ == '__main__':
  print("Running text_utils.py directly:")
  print(f"Capitalized 'hello': {capitalize_string('hello')}")
"""

with open('text_utils.py', 'w') as f:
    f.write(text_content)
print("Created 'text_utils.py'")
```

```python
# 3. Create 'main_application.py' that uses the above modules
main_app_content = """
# Import specific functions from math_operations
from math_operations import add, subtract

# Import the entire text_utils module
import text_utils

print("\n--- Running main_application.py ---")

# Use functions from math_operations
result_add = add(10, 5)
result_subtract = subtract(20, 7)
print(f"Addition result: {result_add}")
print(f"Subtraction result: {result_subtract}")

# Use functions from text_utils
original_text = "python programming"
capitalized_text = text_utils.capitalize_string(original_text)
reversed_text = text_utils.reverse_string(original_text)
print(f"Original text: {original_text}")
print(f"Capitalized text: {capitalized_text}")
print(f"Reversed text: {reversed_text}")

print("-----------------------------------")
"""

with open('main_application.py', 'w') as f:
    f.write(main_app_content)
print("Created 'main_application.py'")
```

Now that we have created these files, we can execute `main_application.py` to see how it integrates the functions from `math_operations.py` and `text_utils.py`.
